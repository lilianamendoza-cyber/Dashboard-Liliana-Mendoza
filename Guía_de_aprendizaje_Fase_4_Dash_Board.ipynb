```python
# Instalar librería para análisis de sentimientos
!pip install vaderSentiment

from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import pandas as pd

# Inicializar el analizador de sentimientos
analyzer = SentimentIntensityAnalyzer()

# Buscar columna de resúmenes (Abstract)
abstract_col = None
for col in df.columns:
    if 'abstract' in col.lower():
        abstract_col = col
        break

if abstract_col:
    print(f"✅ Analizando sentimientos de la columna: '{abstract_col}'\n")

    # Lista para guardar resultados
    sentiment_results = []

    # Analizar cada resumen
    for abstract in df[abstract_col].dropna():
        if len(str(abstract)) > 10:  # Solo si el resumen tiene texto
            sentiment = analyzer.polarity_scores(str(abstract))
            sentiment_results.append(sentiment['compound'])

    # Clasificar sentimientos
    positive = sum(1 for s in sentiment_results if s > 0.05)
    negative = sum(1 for s in sentiment_results if s < -0.05)
    neutral = sum(1 for s in sentiment_results if -0.05 <= s <= 0.05)

    print(f"📊 Resultados del análisis de sentimientos:")
    print(f"   😊 Positivos: {positive} artículos ({positive/len(sentiment_results)*100:.1f}%)")
    print(f"   😞 Negativos: {negative} artículos ({negative/len(sentiment_results)*100:.1f}%)")
    print(f"   😐 Neutrales: {neutral} artículos ({neutral/len(sentiment_results)*100:.1f}%)")
    print(f"   📝 Total analizados: {len(sentiment_results)}")

    # Gráfico de sentimientos
    import matplotlib.pyplot as plt
    labels = ['Positivos', 'Negativos', 'Neutrales']
    sizes = [positive, negative, neutral]
    colors = ['#4CAF50', '#F44336', '#FFC107']

    plt.figure(figsize=(8,6))
    plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
    plt.title('Distribución de Sentimientos en los Resúmenes', fontsize=14)
    plt.axis('equal')
    plt.show()
else:
    print("❌ No se encontró columna de resúmenes (Abstract).")
    print("Columnas disponibles:", df.columns.tolist())
```

```python
# Instalar librería para predicciones
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Usar la columna 'Publication Year' que ya identificamos
year_col = 'Publication Year'

# Contar publicaciones por año
yearly_counts = df[year_col].value_counts().sort_index()

# Preparar datos para regresión lineal
X = yearly_counts.index.values.reshape(-1, 1)  # Años
y = yearly_counts.values  # Número de publicaciones

# Crear y entrenar el modelo
model = LinearRegression()
model.fit(X, y)

# Predecir años futuros (próximos 5 años)
last_year = X[-1][0]
future_years = np.arange(last_year + 1, last_year + 6).reshape(-1, 1)
future_predictions = model.predict(future_years)

# Graficar histórico + predicciones
plt.figure(figsize=(12, 6))
plt.scatter(X, y, color='blue', label='Datos históricos', alpha=0.7)
plt.plot(X, model.predict(X), color='red', linestyle='--', label='Tendencia lineal')
plt.scatter(future_years, future_predictions, color='green', marker='x', s=100, label='Predicciones')
plt.title('Predicción de Publicaciones Científicas', fontsize=14)
plt.xlabel('Año', fontsize=12)
plt.ylabel('Número de Publicaciones', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Mostrar predicciones
print("\n📈 Predicciones para los próximos 5 años:")
for i, year in enumerate(future_years.flatten()):
    print(f"   {year}: {int(future_predictions[i])} publicaciones")

# Mostrar precisión del modelo (R²)
print(f"\n📊 Precisión del modelo (R²): {model.score(X, y):.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

year_col = 'Publication Year'
yearly_counts = df[year_col].value_counts().sort_index()

X = yearly_counts.index.values.reshape(-1, 1)
y = yearly_counts.values

# Usar regresión polinomial (grado 2) que es más realista para crecimiento/declive
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model_poly = LinearRegression()
model_poly.fit(X_poly, y)

# Predecir años futuros
last_year = X[-1][0]
future_years = np.arange(last_year + 1, last_year + 6).reshape(-1, 1)
future_years_poly = poly.transform(future_years)
future_predictions_poly = model_poly.predict(future_years_poly)

# Asegurar que no haya predicciones negativas
future_predictions_poly = np.maximum(future_predictions_poly, 0)

print("📈 Predicciones para los próximos 5 años (Modelo Polinomial):")
for i, year in enumerate(future_years.flatten()):
    print(f"   {year}: {int(future_predictions_poly[i])} publicaciones")

print(f"\n📊 Precisión del modelo polinomial (R²): {model_poly.score(X_poly, y):.3f}")

# Graficar histórico + nueva predicción
plt.figure(figsize=(12, 6))
plt.scatter(X, y, color='blue', label='Datos históricos', alpha=0.7)

# Curva de tendencia polinomial
X_smooth = np.linspace(X.min(), future_years.max(), 100).reshape(-1, 1)
X_smooth_poly = poly.transform(X_smooth)
y_smooth = model_poly.predict(X_smooth_poly)
plt.plot(X_smooth, y_smooth, color='red', linestyle='--', label='Tendencia polinomial')

# Predicciones
plt.scatter(future_years, future_predictions_poly, color='green', marker='o', s=100, label='Predicciones')

plt.title('Predicción de Publicaciones Científicas (Modelo Polinomial)', fontsize=14)
plt.xlabel('Año', fontsize=12)
plt.ylabel('Número de Publicaciones', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================
# DASHBOARD DE INTELIGENCIA ESTRATÉGICA
# NIVEL MAESTRÍA - GESTIÓN DEL CONOCIMIENTO
# AUTORA: LILIANA MENDOZA
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from collections import Counter
import re

# Cargar datos
from google.colab import files
print("📁 Suba el archivo Base de datos (1).xls")
uploaded = files.upload()
df = pd.read_excel('Base de datos (1).xls')
print(f"✅ {len(df)} registros cargados")

# ============================================
# GRÁFICA 1: PUBLICACIONES POR AÑO + ANÁLISIS
# ============================================

year_col = 'Publication Year'
yearly_counts = df[year_col].value_counts().sort_index()

# GRÁFICA
plt.figure(figsize=(12,5))
plt.bar(yearly_counts.index.astype(int), yearly_counts.values, color='steelblue', edgecolor='black')
plt.title('📊 Producción Científica por Año', fontsize=14, fontweight='bold')
plt.xlabel('Año', fontsize=12)
plt.ylabel('Número de Publicaciones', fontsize=12)
plt.xticks(yearly_counts.index.astype(int), rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ANÁLISIS ESTRATÉGICO - GRÁFICA 1
print("\n" + "="*60)
print("📈 ANÁLISIS ESTRATÉGICO - GRÁFICA 1: Producción por Año")
print("="*60)
max_year = yearly_counts.idxmax()
max_count = yearly_counts.max()
min_year = yearly_counts.idxmin()
min_count = yearly_counts.min()
avg_pubs = yearly_counts.mean()
print(f"""
🔎 HALLAZGOS:
   • Año de mayor producción: {int(max_year)} ({max_count} publicaciones)
   • Año de menor producción: {int(min_year)} ({min_count} publicaciones)
   • Promedio anual: {avg_pubs:.1f} publicaciones

💡 IMPLICACIÓN ESTRATÉGICA:
   La producción muestra una tendencia inestable. El pico en {int(max_year)} sugiere
   un interés creciente que no se mantuvo. Se recomienda implementar un programa
   de incentivos a la publicación indexada para estabilizar la producción.
""")

# ============================================
# GRÁFICA 2: ANÁLISIS DE SENTIMIENTOS + ANÁLISIS
# ============================================

analyzer = SentimentIntensityAnalyzer()
abstract_col = None
for col in df.columns:
    if 'abstract' in col.lower():
        abstract_col = col
        break

sentiment_results = []
for abstract in df[abstract_col].dropna():
    if len(str(abstract)) > 10:
        sentiment = analyzer.polarity_scores(str(abstract))
        sentiment_results.append(sentiment['compound'])

positive = sum(1 for s in sentiment_results if s > 0.05)
negative = sum(1 for s in sentiment_results if s < -0.05)
neutral = sum(1 for s in sentiment_results if -0.05 <= s <= 0.05)

# GRÁFICA
plt.figure(figsize=(8,6))
colors_sent = ['#2ecc71', '#e74c3c', '#f39c12']
labels = [f'Positivos ({positive/len(sentiment_results)*100:.1f}%)',
          f'Negativos ({negative/len(sentiment_results)*100:.1f}%)',
          f'Neutrales ({neutral/len(sentiment_results)*100:.1f}%)']
plt.pie([positive, negative, neutral], labels=labels, colors=colors_sent, autopct='%1.1f%%',
        startangle=90, explode=(0.05, 0.05, 0.05), shadow=True)
plt.title('😊 Análisis de Sentimientos en Resúmenes', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ANÁLISIS ESTRATÉGICO - GRÁFICA 2
print("\n" + "="*60)
print("😊 ANÁLISIS ESTRATÉGICO - GRÁFICA 2: Sentimientos")
print("="*60)
print(f"""
🔎 HALLAZGOS:
   • {positive/len(sentiment_results)*100:.1f}% de los artículos son POSITIVOS
   • {negative/len(sentiment_results)*100:.1f}% son NEGATIVOS
   • {neutral/len(sentiment_results)*100:.1f}% son NEUTRALES

💡 IMPLICACIÓN ESTRATÉGICA:
   La abrumadora mayoría de la investigación (casi 90%) tiene un enfoque positivo
   hacia la innovación educativa. Esto representa una oportunidad para posicionar
   a la institución como líder en el área, aprovechando el optimismo del campo.
""")

# ============================================
# GRÁFICA 3: TENDENCIA Y PREDICCIÓN + ANÁLISIS
# ============================================

X = yearly_counts.index.values.reshape(-1, 1)
y = yearly_counts.values

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
model = LinearRegression()
model.fit(X_poly, y)

last_year = X[-1][0]
future_years = np.arange(last_year + 1, last_year + 6).reshape(-1, 1)
future_years_poly = poly.transform(future_years)
predictions = model.predict(future_years_poly)
predictions = np.maximum(predictions, 0)

# GRÁFICA
plt.figure(figsize=(12,5))
plt.scatter(X, y, color='blue', s=80, label='Datos históricos', zorder=5)
X_smooth = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
X_smooth_poly = poly.transform(X_smooth)
y_smooth = model.predict(X_smooth_poly)
plt.plot(X_smooth, y_smooth, color='red', linewidth=2, label='Tendencia polinomial')
plt.scatter(future_years, predictions, color='green', s=120, marker='^', label='Predicción', zorder=5)
plt.title('📈 Tendencia y Predicción de Publicaciones', fontsize=14, fontweight='bold')
plt.xlabel('Año', fontsize=12)
plt.ylabel('Número de Publicaciones', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ANÁLISIS ESTRATÉGICO - GRÁFICA 3
print("\n" + "="*60)
print("📉 ANÁLISIS ESTRATÉGICO - GRÁFICA 3: Tendencia y Predicción")
print("="*60)
r2 = model.score(X_poly, y)
print(f"""
🔎 HALLAZGOS:
   • Precisión del modelo predictivo: R² = {r2:.3f} ({r2*100:.1f}%)
   • La tendencia es DECRECIENTE en los últimos años
   • Proyecciones futuras (sin intervención):
""")
for i, year in enumerate(future_years.flatten()):
    print(f"      - {int(year)}: {int(predictions[i])} publicaciones")
print(f"""
💡 IMPLICACIÓN ESTRATÉGICA:
   Con un R² de {r2*100:.1f}%, el modelo es confiable. La tendencia decreciente
   es una señal de alerta. Se debe activar un plan de acción inmediato para
   revertir esta trayectoria antes de 2028.
""")

# ============================================
# GRÁFICA 4: TOP AUTORES + ANÁLISIS
# ============================================

author_col = None
for col in df.columns:
    if 'author' in col.lower():
        author_col = col
        break

all_authors = []
for authors in df[author_col].dropna():
    authors_list = re.split('[;,]+', str(authors))
    all_authors.extend([a.strip() for a in authors_list if a.strip()])

top_authors = Counter(all_authors).most_common(10)

# GRÁFICA
plt.figure(figsize=(10,6))
names = [a[0][:25] for a in top_authors[:8]]
counts = [a[1] for a in top_authors[:8]]
plt.barh(names[::-1], counts[::-1], color='teal', edgecolor='black')
plt.title('🏆 Autores más Productivos', fontsize=14, fontweight='bold')
plt.xlabel('Número de Publicaciones', fontsize=12)
plt.tight_layout()
plt.show()

# ANÁLISIS ESTRATÉGICO - GRÁFICA 4
print("\n" + "="*60)
print("🏆 ANÁLISIS ESTRATÉGICO - GRÁFICA 4: Top Autores")
print("="*60)
print(f"""
🔎 HALLAZGOS:
   • El autor más productivo tiene {top_authors[0][1]} publicaciones
   • Hay alta concentración: pocos autores generan la mayoría de la producción

💡 IMPLICACIÓN ESTRATÉGICA:
   La alta concentración es un riesgo para la sostenibilidad. Se recomienda:
   1. Programa de mentoría inversa (expertos con noveles)
   2. Incentivos para coautoría entre diferentes grupos
   3. Identificar y potenciar nuevos talentos emergentes
""")

# ============================================
# DASHBOARD RESUMEN EJECUTIVO
# ============================================

print("\n" + "="*70)
print("📋 DASHBOARD RESUMEN EJECUTIVO - INTELIGENCIA ESTRATÉGICA")
print("="*70)
print(f"""
┌─────────────────────────────────────────────────────────────────────┐
│ INDICADOR                          │ VALOR           │ ESTADO      │
├─────────────────────────────────────────────────────────────────────┤
│ Total publicaciones analizadas     │ {len(df)}           │ ✅ Base sólida │
│ Periodo analizado                  │ 2022-2026       │ ✅ 5 años     │
│ Sentimiento positivo               │ {positive/len(sentiment_results)*100:.1f}%           │ ✅ Favorable  │
│ Precisión modelo predictivo (R²)   │ {r2:.3f}            │ ⚠️ Aceptable  │
│ Tendencia                          │ Decreciente     │ 🔴 Alertar    │
│ Concentración autores (top1)       │ {top_authors[0][1]} pubs       │ ⚠️ Riesgo     │
└─────────────────────────────────────────────────────────────────────┘

🎯 RECOMENDACIONES ESTRATÉGICAS PRIORITARIAS:

   1️⃣ FORTALECER SENTIMIENTO POSITIVO (90%)
      → Crear fondo concursable para proyectos de alto impacto

   2️⃣ REVERTIR TENDENCIA DECRECIENTE
      → Implementar incentivos: bonos, reducción de carga docente

   3️⃣ DIVERSIFICAR AUTORES PRODUCTIVOS
      → Redes de colaboración + programa de jóvenes investigadores

   4️⃣ EXPLORAR NUEVAS LÍNEAS TEMÁTICAS
      → IA educativa, gamificación, aprendizaje adaptativo

📊 KPIS DE SEGUIMIENTO ESTRATÉGICO:
   • Proyectos financiados por semestre (meta: 5)
   • Incremento anual de publicaciones (meta: +20%)
   • Nuevos autores con >5 publicaciones (meta: +30%)
   • Nuevas líneas I+D+i incorporadas (meta: 3/año)
""")

print("\n✅ DASHBOARD COMPLETADO - NIVEL MAESTRÍA")
print("📌 Cada gráfica incluye su análisis estratégico correspondiente")
```
